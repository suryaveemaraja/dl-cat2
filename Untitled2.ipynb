```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

print("TensorFlow version:", tf.__version__)
```

```python
df = pd.read_csv("IMDB Dataset.csv")

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

print("Columns:", df.columns.tolist())
print(df.head())

# Keep only the required columns
df = df[["review", "sentiment"]].dropna()
df["sentiment"] = df["sentiment"].astype(str).str.lower().str.strip()

# Remove duplicate reviews
df = df.drop_duplicates(subset=["review"])

# Select 2500 from each class, where available
positive = df[df["sentiment"] == "positive"].sample(
    n=min(2500, (df["sentiment"] == "positive").sum()),
    random_state=42
)

negative = df[df["sentiment"] == "negative"].sample(
    n=min(2500, (df["sentiment"] == "negative").sum()),
    random_state=42
)

df = pd.concat([positive, negative]).sample(
    frac=1, random_state=42
).reset_index(drop=True)

print("Total reviews:", len(df))
print(df["sentiment"].value_counts())
```

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"<br\s*/?>", " ", text)
    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"[^a-z\s']", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["review"] = df["review"].apply(clean_text)

# Convert labels: negative = 0, positive = 1
y = df["sentiment"].map({
    "negative": 0,
    "positive": 1
}).to_numpy()

X_text = df["review"].to_numpy()

print(df.head())
```

```python
X_train_text, X_temp_text, y_train, y_temp = train_test_split(
    X_text, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

X_val_text, X_test_text, y_val, y_test = train_test_split(
    X_temp_text, y_temp,
    test_size=0.50,
    random_state=42,
    stratify=y_temp
)

print("Training:", len(X_train_text))
print("Validation:", len(X_val_text))
print("Testing:", len(X_test_text))
```

```python
X_train_text, X_temp_text, y_train, y_temp = train_test_split(
    X_text, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

X_val_text, X_test_text, y_val, y_test = train_test_split(
    X_temp_text, y_temp,
    test_size=0.50,
    random_state=42,
    stratify=y_temp
)

print("Training:", len(X_train_text))
print("Validation:", len(X_val_text))
print("Testing:", len(X_test_text))
```

```python
MAX_WORDS = 10000
MAX_LEN = 200

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

# Learn vocabulary from training data only
tokenizer.fit_on_texts(X_train_text)

X_train = pad_sequences(
    tokenizer.texts_to_sequences(X_train_text),
    maxlen=MAX_LEN, padding="post", truncating="post"
)

X_val = pad_sequences(
    tokenizer.texts_to_sequences(X_val_text),
    maxlen=MAX_LEN, padding="post", truncating="post"
)

X_test = pad_sequences(
    tokenizer.texts_to_sequences(X_test_text),
    maxlen=MAX_LEN, padding="post", truncating="post"
)

print("Training shape:", X_train.shape)
print("Validation shape:", X_val.shape)
print("Testing shape:", X_test.shape)
```

```python
model = Sequential([
    tf.keras.Input(shape=(MAX_LEN,)),
    Embedding(MAX_WORDS, 64, mask_zero=True),
    LSTM(64),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=8,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)
```

```python
y_prob = model.predict(X_test, verbose=0).ravel()
y_pred = (y_prob >= 0.5).astype(int)

print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall   :", round(recall_score(y_test, y_pred), 4))
print("F1 Score :", round(f1_score(y_test, y_pred), 4))

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred,
    target_names=["Negative", "Positive"]
))
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=["Negative", "Positive"],
    yticklabels=["Negative", "Positive"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("IMDb Review Confusion Matrix")
plt.show()
```

```python
new_reviews = [
    "This movie was fantastic and the acting was excellent.",
    "The film was boring and a complete waste of time.",
    "I loved the story and enjoyed every moment.",
    "Terrible acting and a very disappointing ending.",
    "The visuals were beautiful but the story was weak."
]

new_cleaned = [clean_text(r) for r in new_reviews]

new_seq = tokenizer.texts_to_sequences(new_cleaned)

new_pad = pad_sequences(
    new_seq, maxlen=MAX_LEN,
    padding="post", truncating="post"
)

predictions = model.predict(new_pad, verbose=0).ravel()

for review, prob in zip(new_reviews, predictions):
    label = "Positive" if prob >= 0.5 else "Negative"
    print("\nReview:", review)
    print("Sentiment:", label)
    print("Positive probability:", round(float(prob), 4))
```

```python
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Model Accuracy")
plt.legend()
plt.show()
```
